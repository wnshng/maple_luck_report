`.env`에 `NEXON_OPEN_API_KEY`를 설정한 뒤 실행합니다. API Key는 노트북 출력에 노출하지 마세요.

```python
from datetime import date, timedelta
from dotenv import load_dotenv
import os

load_dotenv('../.env')
api_key = os.getenv('NEXON_OPEN_API_KEY')
bool(api_key)
```

```python
import sys
sys.path.append('..')

from src.nexon_client import NexonMapleClient
from src.preprocess import normalize_cube_history, normalize_starforce_history

client = NexonMapleClient(api_key)
target_date = (date.today() - timedelta(days=1)).isoformat()
cube_payload = client.get_cube_history(target_date, target_date)
starforce_payload = client.get_starforce_history_by_date(target_date)
cube_payload.keys(), starforce_payload.keys()
```

```python
cube_df = normalize_cube_history(cube_payload.get('cube_history', []))
starforce_df = normalize_starforce_history(starforce_payload.get('starforce_history', []))
cube_df.head(), starforce_df.head()
```
